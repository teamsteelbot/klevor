```python
# TRAINING

from yolo import Yolo
from yolo.files import Files
from yolo.train import train_model

# Get YOLO runs folder
yolo_runs_dir = Files.get_yolo_runs_dir_path(Yolo.VERSION_11)

# Get model colab data path
model_colab_data_path = Files.get_model_colab_data_path(Yolo.MODEL_G)

train_model(model='yolo11n.pt', data=model_colab_data_path, epochs=Yolo.EPOCHS,
            imgsz=Yolo.IMAGE_SIZE, project=yolo_runs_dir, name=Yolo.MODEL_G,
            device='0')
```

```python
# RETRAINING

from yolo.files import Files
from yolo.train import train_model
from yolo import Yolo

# Get YOLO runs folder
yolo_runs_dir = Files.get_yolo_runs_dir_path(Yolo.VERSION_11)

# Get model colab data path
model_colab_data_path = Files.get_model_colab_data_path(Yolo.MODEL_G)

# Get the best PyTorch model
model = Files.get_model_best_pt_path(Yolo.MODEL_G, Yolo.VERSION_11)

# Train model
train_model(model=model, data=model_colab_data_path, epochs=Yolo.EPOCHS,
            imgsz=Yolo.IMAGE_SIZE, project=yolo_runs_dir, name=Yolo.MODEL_G,
            device='0')
```
